## Langkah 4 - Training dan Evaluasi Model

```python
import pandas as pd
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

# 1. Load data original kembali untuk menghindari nilai NaN yang sudah terlanjur tersimpan
df_temp = pd.read_csv('spam.csv', encoding='latin-1')
df_temp = df_temp.drop(df_temp.iloc[:, 2:], axis=1)
df_temp = df_temp.rename(columns={'v1': 'Labels', 'v2': 'SMS'})

# 2. Bersihkan label dan mapping dengan aman
df_temp['Labels'] = df_temp['Labels'].astype(str).str.strip().str.lower()
new_labels = {'spam': 1, 'ham': 0}
df_temp['Labels'] = df_temp['Labels'].map(new_labels)

X = df_temp['SMS'].values
y = df_temp['Labels'].values

# 3. Split data training dan testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# 4. Inisiasi dan transform CountVectorizer baru
bow = CountVectorizer()
X_train = bow.fit_transform(X_train)
X_test = bow.transform(X_test)

# 5. Inisiasi dan Fit model Naive Bayes
mnb = MultinomialNB()
mnb.fit(X_train, y_train)

# Prediksi dan Evaluasi
y_pred_train = mnb.predict(X_train)
acc_train = accuracy_score(y_train, y_pred_train)

y_pred_test = mnb.predict(X_test)
acc_test = accuracy_score(y_test, y_pred_test)

print(f'Hasil akurasi data train: {acc_train}')
print(f'Hasil akurasi data test: {acc_test}')
```
